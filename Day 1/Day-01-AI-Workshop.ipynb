## 📌 Section 1: What is Artificial Intelligence?

**Artificial Intelligence (AI)** refers to the simulation of human intelligence in machines that are programmed to think, reason, learn, and solve problems.

### Key Definitions
| Term | Definition |
|------|------------|
| Intelligence | The ability to acquire and apply knowledge and skills |
| Artificial | Made or produced by human beings rather than occurring naturally |
| AI | Computer systems that perform tasks that normally require human intelligence |

### Goals of AI
- 🧠 **Reasoning** — Drawing conclusions from data
- 📚 **Learning** — Improving from experience
- 👁️ **Perception** — Understanding images, speech, and text
- 🗣️ **Natural Language Processing** — Understanding and generating human language
- 🤸 **Problem-Solving** — Finding solutions to complex problems

## 📌 Section 2: Types of AI

AI is broadly classified based on **capabilities** and **functionality**.

### 🔹 Based on Capabilities

| Type | Description | Example |
|------|-------------|--------|
| **Narrow AI (Weak AI)** | Designed for a specific task | Siri, Chess engines, Spam filters |
| **General AI (Strong AI)** | Human-like intelligence across all tasks | Hypothetical — doesn't exist yet |
| **Super AI** | Surpasses human intelligence in every domain | Theoretical/Future concept |

### 🔹 Based on Functionality

| Type | Description | Example |
|------|-------------|--------|
| **Reactive Machines** | No memory; reacts to current input | IBM Deep Blue (Chess) |
| **Limited Memory** | Uses past data for decisions | Self-driving cars |
| **Theory of Mind** | Understands emotions/beliefs | Research stage |
| **Self-Aware AI** | Has consciousness | Theoretical |

> 🧠 **Today, all real-world AI is Narrow AI!**

````python
# 🔍 Let's Explore Types of AI with Python

ai_types = {
    "Narrow AI": ["Siri", "Alexa", "Google Translate", "Spam Filters", "Recommendation Systems"],
    "General AI": ["Hypothetical — Not yet achieved"],
    "Super AI": ["Theoretical — Future concept"]
}

print("📊 Types of AI and Examples\n")
print("-" * 50)

for ai_type, examples in ai_types.items():
    print(f"\n🔹 {ai_type}:")
    for example in examples:
        print(f"   • {example}")

print("-" * 50)
print("\n✅ Currently, we only have Narrow AI in practice!")
````

## 📌 Section 3: AI vs Machine Learning vs Deep Learning

These three terms are often confused. Let's clarify them:

```
┌─────────────────────────────────────────┐
│              ARTIFICIAL INTELLIGENCE     │
│  ┌───────────────────────────────────┐   │
│  │        MACHINE LEARNING           │   │
│  │  ┌─────────────────────────────┐  │   │
│  │  │       DEEP LEARNING         │  │   │
│  │  └─────────────────────────────┘  │   │
│  └───────────────────────────────────┘   │
└─────────────────────────────────────────┘
```

### Key Differences

| Feature | AI | Machine Learning | Deep Learning |
|---------|-----|-------------------|---------------|
| **Definition** | Broad field of intelligent machines | Subset of AI; learns from data | Subset of ML; uses neural networks |
| **Data Required** | Varies | Moderate | Large amounts |
| **Human Intervention** | High | Medium | Low |
| **Examples** | Chatbots, Robotics | Linear Regression, Decision Trees | Image Recognition, GPT |
| **Hardware** | Varies | CPU sufficient | GPU/TPU required |

### Simple Analogy 🍕
> - **AI** = Making food (broad goal)
> - **ML** = Following a recipe (learning from data/instructions)
> - **DL** = Having taste buds (automatic feature detection)

````python
# 📊 Visualizing AI > ML > Deep Learning relationship
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(1, 1, figsize=(8, 6))

# Draw nested circles
colors = ['#AED6F1', '#85C1E9', '#2E86C1']
labels = ['Artificial Intelligence', 'Machine Learning', 'Deep Learning']
sizes = [5000, 3000, 1000]

for size, color, label in zip(sizes, colors, labels):
    circle = plt.Circle((0.5, 0.5), size/10000, color=color, alpha=0.8)
    ax.add_patch(circle)

# Add labels
ax.text(0.5, 0.85, 'Artificial Intelligence', ha='center', va='center',
        fontsize=14, fontweight='bold', color='#1A252F')
ax.text(0.5, 0.68, 'Machine Learning', ha='center', va='center',
        fontsize=12, fontweight='bold', color='#1A252F')
ax.text(0.5, 0.50, 'Deep\nLearning', ha='center', va='center',
        fontsize=10, fontweight='bold', color='white')

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
ax.set_facecolor('#F8F9FA')
fig.patch.set_facecolor('#F8F9FA')
plt.title('AI ⊃ Machine Learning ⊃ Deep Learning', fontsize=15, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
print("\n✅ Deep Learning is a subset of ML, which is a subset of AI!")
````

## 📌 Section 4: Real-World AI Applications

AI is already transforming industries all around us!

| Domain | Application | Tool/Company |
|--------|-------------|-------------|
| 🏥 **Healthcare** | Disease diagnosis, Drug discovery | DeepMind, IBM Watson |
| 🚗 **Transportation** | Self-driving cars, Route optimization | Tesla, Waymo |
| 💰 **Finance** | Fraud detection, Stock prediction | PayPal, JPMorgan |
| 🛒 **E-commerce** | Product recommendations, Chatbots | Amazon, Alibaba |
| 🎓 **Education** | Personalized learning, Auto-grading | Coursera, Duolingo |
| 🌾 **Agriculture** | Crop disease detection, Yield prediction | Plantix, IBM |
| 🎮 **Gaming** | NPC behavior, Game generation | OpenAI Five, AlphaGo |
| 🗣️ **Language** | Translation, Summarization, Chatbots | Google Translate, ChatGPT |

````python
# 📊 Visualize AI Application Areas

import matplotlib.pyplot as plt
import numpy as np

# AI adoption levels (hypothetical % adoption) across industries
industries = ['Healthcare', 'Finance', 'E-commerce', 'Transportation',
              'Education', 'Agriculture', 'Manufacturing', 'Entertainment']

adoption_rates = [72, 85, 91, 65, 58, 45, 78, 88]

colors = plt.cm.Blues(np.linspace(0.4, 0.9, len(industries)))

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(industries, adoption_rates, color=colors, edgecolor='white', linewidth=0.8)

# Add value labels
for bar, val in zip(bars, adoption_rates):
    ax.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
            f'{val}%', va='center', fontsize=11, fontweight='bold', color='#2C3E50')

ax.set_xlabel('AI Adoption Rate (%)', fontsize=12)
ax.set_title('🤖 AI Adoption Across Industries', fontsize=15, fontweight='bold', pad=15)
ax.set_xlim(0, 105)
ax.spines[['top','right','left']].set_visible(False)
ax.tick_params(left=False)
ax.set_facecolor('#F8F9FA')
fig.patch.set_facecolor('#F8F9FA')
plt.tight_layout()
plt.show()
````

### 📈 AI Market Growth
> The global AI market is expected to exceed **$1.8 trillion by 2030**, growing at ~37% annually.

````python
# 📈 AI Market Growth Projection

import matplotlib.pyplot as plt
import numpy as np

years = [2020, 2021, 2022, 2023, 2024, 2025, 2026, 2027, 2028, 2029, 2030]
market_size = [62, 86, 120, 165, 230, 320, 445, 620, 860, 1200, 1800]  # in billion USD

fig, ax = plt.subplots(figsize=(10, 5))

ax.fill_between(years, market_size, alpha=0.2, color='#2E86C1')
ax.plot(years, market_size, color='#2E86C1', linewidth=2.5, marker='o',
        markersize=7, markerfacecolor='white', markeredgewidth=2)

# Annotate key points
for x, y in zip(years[::2], market_size[::2]):
    ax.annotate(f'${y}B', (x, y), textcoords='offset points',
                xytext=(0, 12), ha='center', fontsize=9, color='#1A252F', fontweight='bold')

ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Market Size (Billion USD)', fontsize=12)
ax.set_title('📈 Global AI Market Growth (2020–2030)', fontsize=14, fontweight='bold', pad=15)
ax.spines[['top','right']].set_visible(False)
ax.set_facecolor('#F8F9FA')
fig.patch.set_facecolor('#F8F9FA')
plt.tight_layout()
plt.show()
print("\n✅ The AI industry is one of the fastest growing sectors globally!")
````
